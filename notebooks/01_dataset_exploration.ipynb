### 3. Summary of Stimulation Conditions
In the standardized Visual Coding protocol:
- **Primary condition**: Single 10-ms square optical pulses (varied power levels, 30-50 trials/level)
- **Secondary conditions**: 5-ms pulses, 2.5-ms pulse trains (10 Hz for 1 s), 1-s raised cosine ramps.
